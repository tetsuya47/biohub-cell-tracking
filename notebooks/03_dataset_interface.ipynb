# Milestone 3 — Dataset Interface

This notebook verifies the unified `BiohubDataset` interface that combines Zarr images and GEFF ground truth lineage graphs.

## 1. Load BiohubDataset

```python
import sys
from pathlib import Path

# Add project root to sys.path to allow importing src
project_root = str(Path("..").resolve())
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.datasets.biohub_dataset import BiohubDataset

# Locate a matching pair of Zarr/GEFF files
train_dir = Path("../data/raw/train")
zarr_paths = sorted(list(train_dir.glob("*.zarr")))
geff_paths = sorted(list(train_dir.glob("*.geff")))

test_zarr_path = None
test_geff_path = None
for z_path in zarr_paths:
    g_path = z_path.with_suffix(".geff")
    if g_path in geff_paths:
        test_zarr_path = z_path
        test_geff_path = g_path
        break

if not test_zarr_path or not test_geff_path:
    raise FileNotFoundError("Could not find matching Zarr/GEFF pair")

dataset = BiohubDataset(test_zarr_path, test_geff_path)
print(f"Loaded dataset with Zarr={test_zarr_path.name} and GEFF={test_geff_path.name}")
```

## 2. Print Dataset Length

```python
print(f"Dataset length: {len(dataset)}")
```
