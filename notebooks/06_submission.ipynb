# 06. Pipeline and Submission Generation

This notebook runs the unified cell tracking inference pipeline and serializes outputs into the final submission CSV file.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from src.submission.writer import SubmissionWriter
from src.inference.pipeline import InferencePipeline
from src.models.detection.base import BaseDetector
from src.models.tracking.base import BaseTracker
from src.models.division.base import BaseDivisionDetector
```

## 1. Initialize Pipeline and Load Models

```python
detector = BaseDetector()
tracker = BaseTracker()
division_detector = BaseDivisionDetector()

pipeline = InferencePipeline(detector, tracker, division_detector)
print("Inference Pipeline successfully assembled.")
```

## 2. Run Inference and Generate CSV

```python
# Create synthetic temporal volume: 5 timesteps, 1 channel, depth=8, H=64, W=64
volume = np.random.randn(5, 1, 8, 64, 64)

writer = pipeline.run_inference(volume)
csv_out_path = Path("submission_example.csv")
writer.save_csv(csv_out_path)

df = pd.read_csv(csv_out_path)
print(f"Generated {len(df)} rows. Head:")
print(df.head())
```
