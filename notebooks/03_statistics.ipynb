# 03. Dataset Statistics

This notebook computes statistical summaries for temporal cell volumes, aiding normalization scaling and model preprocessing setup.

```python
import numpy as np
from src.datasets.statistics import compute_dataset_stats
```

## Compute Statistics on a Sample Volume

```python
# Generate synthetic distribution mimicking cell image channels
sample_vol = np.random.normal(loc=100.0, scale=15.0, size=(16, 64, 64))
sample_vol = np.clip(sample_vol, 0, 255)

stats = compute_dataset_stats(sample_vol)
for key, val in stats.items():
    print(f"{key.capitalize()}: {val:.4f}")
```
