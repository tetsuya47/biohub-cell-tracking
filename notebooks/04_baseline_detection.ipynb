# 04. Baseline Detection Model

This notebook covers the baseline model for 3D cell centroid and segmentation detection, validating network input/output shapes and the loss wrapper.

```python
import torch
from src.models.detection.base import BaseDetector
from src.losses.detection import DetectionLoss
```

## 1. Verify Detector Shape and Forward Pass

```python
detector = BaseDetector(in_channels=1, out_channels=1)
# Create synthetic batch: size=1, channels=1, depth=8, height=64, width=64
x = torch.randn(1, 1, 8, 64, 64)
y_pred = detector(x)
print(f"Input shape:  {list(x.shape)}")
print(f"Output shape: {list(y_pred.shape)}")
```

## 2. Test Combined Detection Loss (BCE + Dice)

```python
loss_fn = DetectionLoss(dice_weight=0.5)
# Mock binary ground truth targets
y_true = torch.randint(0, 2, y_pred.shape).float()
loss = loss_fn(y_pred, y_true)
print(f"Computed detection loss: {loss.item():.6f}")
```
