# Milestone 6 — PyTorch DataLoader

This notebook verifies the training DataLoader builder `BiohubDataLoader` configured with PyTorch's default collation.

## 1. Load BiohubDataLoader

```python
import sys
from pathlib import Path
import torch

# Add project root to sys.path to allow importing src
project_root = str(Path("..").resolve())
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.datasets.dataloader import BiohubDataLoader
print("BiohubDataLoader loaded successfully.")
```

## 2. Create DataLoader

```python
# Find matching dataset pair programmatically
train_dir = Path("../data/raw/train")
zarr_paths = sorted(list(train_dir.glob("*.zarr")))
geff_paths = sorted(list(train_dir.glob("*.geff")))

test_zarr_path = None
test_geff_path = None

for z_path in zarr_paths:
    g_path = z_path.with_suffix(".geff")
    if g_path in geff_paths:
        test_zarr_path = z_path
        test_geff_path = g_path
        break

if not test_zarr_path or not test_geff_path:
    raise FileNotFoundError("Matching Zarr/GEFF pair not found")

bh_loader = BiohubDataLoader(
    test_zarr_path,
    test_geff_path,
    batch_size=1,
    shuffle=False,
)
loader = bh_loader.get_train_loader()
```

## 3. Load First Batch

```python
batch = next(iter(loader))
```

## 4. Print Batch Keys & Batch Size

```python
print(f"Batch keys: {list(batch.keys())}")
print(f"Batch size: {len(batch['image'])}")
```

## 5. Print Tensor Shapes

```python
print(f"image tensor shape:     {batch['image'].shape}")
print(f"node_ids tensor shape:  {batch['node_ids'].shape}")
print(f"positions tensor shape: {batch['positions'].shape}")
print(f"edges tensor shape:     {batch['edges'].shape}")
```
