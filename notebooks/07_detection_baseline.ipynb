```python
import torch
from src.models.detection.baseline_detector import BaselineDetector
from src.losses.detection_loss import DetectionLoss
from src.preprocessing.target_generator import TargetGenerator
from src.postprocessing.peak_detector import PeakDetector
```

```python
# 1. Instantiate model
model = BaselineDetector()
```

```python
# 2. Create random tensor
x = torch.randn(2, 1, 8, 32, 32)
```

```python
# 3. Forward pass
output = model(x)
```

```python
# 4. Print output shape
print("Output shape:", output.shape)
```

```python
# 5. Instantiate target generator and peak detector
target_gen = TargetGenerator(volume_shape=(8, 32, 32))
peak_det = PeakDetector(threshold=0.5)
```

```python
# 6. Run target generator and peak detector computations
positions = torch.randn(5, 4)
target_heatmap = target_gen.generate(positions)
detected_coordinates = peak_det.detect(output)
print("Target heatmap shape:", target_heatmap.shape)
print("Detected coordinates shape:", detected_coordinates.shape)
```

```python
# 7. Instantiate loss
loss_fn = DetectionLoss()
```

```python
# 8. Run one loss computation
# Replicate target heatmap to match prediction batch dimension for demonstration
targets = torch.cat([target_heatmap.unsqueeze(0)] * output.shape[0], dim=0)
loss = loss_fn(output, targets)
print("Computed loss:", loss.item())
```
