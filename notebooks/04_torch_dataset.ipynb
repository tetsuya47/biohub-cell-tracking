# Milestone 4 — PyTorch Dataset

This notebook verifies the PyTorch dataset wrapper `BiohubTorchDataset` which yields PyTorch tensors from matching image and graph datasets.

## 1. Import BiohubTorchDataset

```python
import sys
from pathlib import Path
import torch

# Add project root to sys.path to allow importing src
project_root = str(Path("..").resolve())
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.datasets.torch_dataset import BiohubTorchDataset
print("BiohubTorchDataset imported successfully.")
```

## 2. Load Dataset

```python
# Find matching dataset pair programmatically
train_dir = Path("../data/raw/train")
zarr_paths = sorted(list(train_dir.glob("*.zarr")))
geff_paths = sorted(list(train_dir.glob("*.geff")))

test_zarr_path = None
test_geff_path = None

for z_path in zarr_paths:
    g_path = z_path.with_suffix(".geff")
    if g_path in geff_paths:
        test_zarr_path = z_path
        test_geff_path = g_path
        break

if not test_zarr_path or not test_geff_path:
    raise FileNotFoundError("Matching Zarr/GEFF pair not found")

print(f"Loading dataset with files:\n  Image: {test_zarr_path.name}\n  Graph: {test_geff_path.name}")
dataset = BiohubTorchDataset(test_zarr_path, test_geff_path)
```

## 3. Print Dataset Length

```python
print(f"Dataset length: {len(dataset)}")
```

## 4. Load Sample 0

```python
sample = dataset[0]
```

## 5. Print Tensor Dtypes

```python
print("Tensor dtypes:")
for key, val in sample.items():
    print(f"  {key}: {val.dtype}")
```

## 6. Print Tensor Shapes

```python
print("Tensor shapes:")
for key, val in sample.items():
    print(f"  {key}: {val.shape}")
```
