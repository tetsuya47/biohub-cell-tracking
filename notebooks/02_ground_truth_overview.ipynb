# Milestone 2 — Read Biohub Ground Truth (.geff)

This notebook performs a basic overview of the Biohub cell lineage ground truth data stored in `.geff` format using the custom `GEFFReader` class.

## 1. Load Libraries

```python
import sys
from pathlib import Path

# Add project root to sys.path to allow importing src
project_root = str(Path("..").resolve())
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.datasets.geff_reader import GEFFReader
print("Libraries loaded successfully.")
```

## 2. Load One .geff Dataset

```python
train_dir = Path("../data/raw/train")
geff_paths = sorted(list(train_dir.glob("*.geff")))

if not geff_paths:
    raise FileNotFoundError("No .geff files found in data/raw/train/")

first_geff_path = geff_paths[0]
print(f"Selected ground truth dataset: {first_geff_path.name}")

reader = GEFFReader(first_geff_path)
```

## 3. Print Node Count & Edge Count

```python
node_ids = reader.get_node_ids()
edges = reader.get_edges()

print(f"Node count: {len(node_ids)}")
print(f"Edge count: {len(edges)}")
```

## 4. Display First 10 Node IDs

```python
print("First 10 node IDs:")
print(node_ids[:10])
```

## 5. Display First 10 Edges

```python
print("First 10 edges (source, target):")
print(edges[:10])
```

## 6. Print Summary

```python
summary_dict = reader.summary()
for key, value in summary_dict.items():
    print(f"{key}: {value}")
```
